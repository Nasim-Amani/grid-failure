```python
# ==========================================
# Step 4: Jalali Date Conversion + Time Features
# ==========================================

!pip install persiantools -q

import pandas as pd
from persiantools.jdatetime import JalaliDate


# Work on current dataframe
df = df.copy()


# ------------------------------------------
# 1. Convert Jalali date to Gregorian date
# ------------------------------------------

def convert_jalali_date(date):

    try:
        y, m, d = map(
            int,
            str(date).split("/")
        )

        return JalaliDate(
            y, m, d
        ).to_gregorian()

    except:
        return pd.NaT


df["date_gregorian"] = (
    df["تاریخ اطلاع"]
    .apply(convert_jalali_date)
)



# ------------------------------------------
# 2. Create datetime
# ------------------------------------------

df["datetime"] = (
    df["date_gregorian"].astype(str)
    + " "
    + df["ساعت اطلاع"].astype(str)
)


df["datetime"] = pd.to_datetime(
    df["datetime"],
    errors="coerce"
)



# ------------------------------------------
# 3. Gregorian calendar features
# ------------------------------------------

df["gregorian_year"] = (
    df["datetime"].dt.year
)

df["gregorian_month"] = (
    df["datetime"].dt.month
)

df["gregorian_day"] = (
    df["datetime"].dt.day
)

df["hour"] = (
    df["datetime"].dt.hour
)



# ------------------------------------------
# 4. Iranian weekday features
# ------------------------------------------

# Pandas:
# 0 Monday ... 6 Sunday

df["weekday"] = (
    df["datetime"].dt.weekday
)


# Iran:
# 0 Saturday ... 6 Friday

df["iran_weekday"] = (
    (df["weekday"] + 2) % 7
)


iran_weekday_names = {

    0: "Saturday",
    1: "Sunday",
    2: "Monday",
    3: "Tuesday",
    4: "Wednesday",
    5: "Thursday",
    6: "Friday"
}


df["iran_weekday_name"] = (
    df["iran_weekday"]
    .map(iran_weekday_names)
)



# ------------------------------------------
# 5. Jalali calendar features
# ------------------------------------------

def extract_jalali_parts(jalali_date):

    try:

        y, m, d = map(
            int,
            str(jalali_date).split("/")
        )

        return pd.Series(
            [y, m, d]
        )

    except:

        return pd.Series(
            [None, None, None]
        )


df[
    [
        "jalali_year",
        "jalali_month",
        "jalali_day"
    ]
] = (
    df["تاریخ اطلاع"]
    .apply(extract_jalali_parts)
)



# ------------------------------------------
# 6. Jalali season
# ------------------------------------------

def get_jalali_season(jalali_month):

    if jalali_month in [1, 2, 3]:
        return "Spring"

    elif jalali_month in [4, 5, 6]:
        return "Summer"

    elif jalali_month in [7, 8, 9]:
        return "Autumn"

    elif jalali_month in [10, 11, 12]:
        return "Winter"

    else:
        return "Unknown"


df["jalali_season"] = (
    df["jalali_month"]
    .apply(get_jalali_season)
)



# ------------------------------------------
# 7. Check result
# ------------------------------------------

print("Jalali season distribution:")

display(
    df["jalali_season"]
    .value_counts()
)


print("\nDate conversion examples:")

display(
    df[
        [
            "تاریخ اطلاع",
            "ساعت اطلاع",
            "date_gregorian",
            "datetime",
            "gregorian_year",
            "gregorian_month",
            "jalali_year",
            "jalali_month",
            "jalali_season",
            "hour",
            "iran_weekday_name"
        ]
    ].head()
)


print("\nDatetime missing values:")
print(
    df["datetime"]
    .isna()
    .sum()
)


print("\nNew dataframe shape:")
print(df.shape)
```

## Daily Feeder Aggregation Strategy

The original outage dataset contains event-level records (one row = one outage event).
To create a feeder-day health dataset, outage events occurring on the same feeder and the same day are aggregated.

Different aggregation methods are selected based on the physical meaning of each feature:

| Feature | Aggregation Method | Meaning in Daily Feeder Health Dataset |
|---|---|---|
| unplanned_failures | SUM | Total number of unexpected failures occurring on the feeder during the day. Represents failure frequency. |
| total_outages | COUNT | Total number of outage events recorded during the day (planned + unplanned). Represents overall outage activity. |
| planned_outages | Conditional SUM | Number of planned outages during the day. Separates maintenance activity from unexpected failures. |
| total_duration | SUM | Total interruption time accumulated during the day. Represents overall outage burden. |
| max_duration | MAX | Duration of the most severe single outage event. Captures extreme outage severity. |
| total_energy_loss | SUM | Total energy not supplied during the day. Represents cumulative customer impact. |
| max_energy_loss | MAX | Highest energy loss caused by a single outage event. Identifies extreme impact events. |
| average_current | MEAN | Average interrupted current level across outage events. Represents typical electrical stress. |
| max_interrupted_load | MAX | Maximum interrupted load among events. Captures peak network impact. |
| switching_events | COUNT | Number of outage records with a recorded switching device. Represents operational activity. |

Aggregation principle:

- **SUM** → used for cumulative effects (failures, duration, energy loss).
- **MAX** → used for extreme events (worst outage severity and impact).
- **MEAN** → used for continuous measurements where the average condition is meaningful.
- **COUNT** → used for event frequency and operational activity.

Keeping both cumulative and extreme indicators allows the model to distinguish between:

- feeders with frequent small disturbances,
- feeders with rare but highly damaging failures.

```python
# ==========================================
# Daily Numerical Feeder Aggregation
# ==========================================

import pandas as pd


# ------------------------------------------
# Create daily date
# ------------------------------------------

df_reference["date"] = (
    df_reference["datetime"]
    .dt.date
)


# ------------------------------------------
# Create daily feeder health numerical features
# ------------------------------------------

daily_numeric = (
    df_reference
    .groupby(
        [
            "mv_feeder_gis",
            "date"
        ]
    )
    .agg(

        # Failure behavior

        unplanned_failures=(
            "failure_target",
            "sum"
        ),

        total_outages=(
            "failure_target",
            "count"
        ),

        planned_outages=(
            "failure_target",
            lambda x: (x == 0).sum()
        ),


        # Impact

        total_duration=(
            "outage_minutes",
            "sum"
        ),

        max_duration=(
            "outage_minutes",
            "max"
        ),

        total_energy_loss=(
            "energy_not_supplied_kwh",
            "sum"
        ),

        max_energy_loss=(
            "energy_not_supplied_kwh",
            "max"
        ),


        # Electrical stress

        average_current=(
            "initial_current",
            "mean"
        ),

        max_interrupted_load=(
            "interrupted_load_mwh",
            "max"
        ),


        # Operational activity

        switching_events=(
            "switch_device",
            "count"
        )

    )
    .reset_index()
)


print("Daily numerical dataset shape:")
print(daily_numeric.shape)


display(
    daily_numeric.head()
)
```

```python
# ==========================================
# Sort feeder history chronologically
# ==========================================

daily_numeric = (
    daily_numeric
    .sort_values(
        [
            "mv_feeder_gis",
            "date"
        ]
    )
    .reset_index(drop=True)
)


print("Sorted successfully.")

display(
    daily_numeric.head()
)
```
