```python
# ==========================================
# Step 4: Jalali Date Conversion + Time Features
# ==========================================

!pip install persiantools -q

import pandas as pd
from persiantools.jdatetime import JalaliDate


df = df.copy()


# ------------------------------------------
# 1. Convert Jalali date to Gregorian
# ------------------------------------------

def convert_jalali_date(date):

    try:
        y, m, d = map(
            int,
            str(date).split("/")
        )

        return JalaliDate(
            y, m, d
        ).to_gregorian()

    except Exception:
        return pd.NaT


df["date_gregorian"] = pd.to_datetime(
    df["تاریخ اطلاع"].apply(convert_jalali_date),
    errors="coerce"
)


# ------------------------------------------
# 2. Create datetime
# ------------------------------------------

df["datetime"] = (
    df["date_gregorian"]
    +
    pd.to_timedelta(
        df["ساعت اطلاع"].astype(str),
        errors="coerce"
    )
)


# ------------------------------------------
# 3. Gregorian calendar features
# ------------------------------------------

df["gregorian_year"] = df["datetime"].dt.year
df["gregorian_month"] = df["datetime"].dt.month
df["gregorian_day"] = df["datetime"].dt.day

df["hour"] = df["datetime"].dt.hour


# ------------------------------------------
# 4. Iranian weekday
# ------------------------------------------

# Pandas:
# Monday = 0 ... Sunday = 6

df["weekday"] = df["datetime"].dt.weekday


# Iran:
# Saturday = 0 ... Friday = 6

df["iran_weekday"] = (
    (df["weekday"] + 2) % 7
)


iran_weekday_names = {
    0: "Saturday",
    1: "Sunday",
    2: "Monday",
    3: "Tuesday",
    4: "Wednesday",
    5: "Thursday",
    6: "Friday"
}


df["iran_weekday_name"] = (
    df["iran_weekday"]
    .map(iran_weekday_names)
)


# ------------------------------------------
# 5. Jalali calendar features
# ------------------------------------------

def extract_jalali_parts(jalali_date):

    try:

        y, m, d = map(
            int,
            str(jalali_date).split("/")
        )

        return pd.Series(
            [y, m, d]
        )

    except Exception:

        return pd.Series(
            [pd.NA, pd.NA, pd.NA]
        )


df[
    [
        "jalali_year",
        "jalali_month",
        "jalali_day"
    ]
] = (
    df["تاریخ اطلاع"]
    .apply(extract_jalali_parts)
)


# ------------------------------------------
# 6. Jalali season
# ------------------------------------------

def get_jalali_season(month):

    if month in [1, 2, 3]:
        return "Spring"

    elif month in [4, 5, 6]:
        return "Summer"

    elif month in [7, 8, 9]:
        return "Autumn"

    elif month in [10, 11, 12]:
        return "Winter"

    return "Unknown"


df["jalali_season"] = (
    df["jalali_month"]
    .apply(get_jalali_season)
)


# ------------------------------------------
# 7. Check result
# ------------------------------------------

print("Jalali season distribution:")

display(
    df["jalali_season"]
    .value_counts(dropna=False)
)


print("\nDate conversion examples:")

display(
    df[
        [
            "تاریخ اطلاع",
            "ساعت اطلاع",
            "date_gregorian",
            "datetime",
            "gregorian_year",
            "gregorian_month",
            "jalali_year",
            "jalali_month",
            "jalali_season",
            "hour",
            "iran_weekday_name"
        ]
    ].head()
)


print("\nDatetime missing values:")
print(
    df["datetime"]
    .isna()
    .sum()
)


print("\nNew dataframe shape:")
print(df.shape)
```

## Daily Feeder Aggregation Strategy

The original outage dataset contains event-level records (one row = one outage event).
To create a feeder-day health dataset, outage events occurring on the same feeder and the same day are aggregated.

Different aggregation methods are selected based on the physical meaning of each feature:

| Feature | Aggregation Method | Meaning in Daily Feeder Health Dataset |
|---|---|---|
| unplanned_failures | SUM | Total number of unexpected failures occurring on the feeder during the day. Represents failure frequency. |
| total_outages | COUNT | Total number of outage events recorded during the day (planned + unplanned). Represents overall outage activity. |
| planned_outages | Conditional SUM | Number of planned outages during the day. Separates maintenance activity from unexpected failures. |
| total_duration | SUM | Total interruption time accumulated during the day. Represents overall outage burden. |
| max_duration | MAX | Duration of the most severe single outage event. Captures extreme outage severity. |
| total_energy_loss | SUM | Total energy not supplied during the day. Represents cumulative customer impact. |
| max_energy_loss | MAX | Highest energy loss caused by a single outage event. Identifies extreme impact events. |
| average_current | MEAN | Average interrupted current level across outage events. Represents typical electrical stress. |
| max_interrupted_load | MAX | Maximum interrupted load among events. Captures peak network impact. |
| switching_events | COUNT | Number of outage records with a recorded switching device. Represents operational activity. |

Aggregation principle:

- **SUM** → used for cumulative effects (failures, duration, energy loss).
- **MAX** → used for extreme events (worst outage severity and impact).
- **MEAN** → used for continuous measurements where the average condition is meaningful.
- **COUNT** → used for event frequency and operational activity.

Keeping both cumulative and extreme indicators allows the model to distinguish between:

- feeders with frequent small disturbances,
- feeders with rare but highly damaging failures.

```python
# ==========================================
# Daily Numerical Feeder Aggregation
# ==========================================

import pandas as pd


# ------------------------------------------
# 1. Create daily date
# ------------------------------------------

df_reference["date"] = (
    df_reference["datetime"]
    .dt.normalize()
)


# ------------------------------------------
# 2. Separate planned / unplanned impact
# ------------------------------------------

df_reference["unplanned_duration"] = (
    df_reference["outage_minutes"]
    .where(
        df_reference["failure_target"] == 1,
        0
    )
)

df_reference["planned_duration"] = (
    df_reference["outage_minutes"]
    .where(
        df_reference["failure_target"] == 0,
        0
    )
)


df_reference["unplanned_energy_loss"] = (
    df_reference["energy_not_supplied_kwh"]
    .where(
        df_reference["failure_target"] == 1,
        0
    )
)

df_reference["planned_energy_loss"] = (
    df_reference["energy_not_supplied_kwh"]
    .where(
        df_reference["failure_target"] == 0,
        0
    )
)


# ------------------------------------------
# 3. Daily numerical aggregation
# ------------------------------------------

daily_numeric = (
    df_reference
    .groupby(
        [
            "mv_feeder_gis",
            "date"
        ]
    )
    .agg(

        # Failure frequency
        unplanned_failures=(
            "failure_target",
            "sum"
        ),

        total_outages=(
            "failure_target",
            "count"
        ),

        planned_outages=(
            "failure_target",
            lambda x: (x == 0).sum()
        ),


        # Duration
        total_duration=(
            "outage_minutes",
            "sum"
        ),

        unplanned_duration=(
            "unplanned_duration",
            "sum"
        ),

        planned_duration=(
            "planned_duration",
            "sum"
        ),

        max_duration=(
            "outage_minutes",
            "max"
        ),


        # Energy
        total_energy_loss=(
            "energy_not_supplied_kwh",
            "sum"
        ),

        unplanned_energy_loss=(
            "unplanned_energy_loss",
            "sum"
        ),

        planned_energy_loss=(
            "planned_energy_loss",
            "sum"
        ),

        max_energy_loss=(
            "energy_not_supplied_kwh",
            "max"
        ),


        # Electrical stress
        average_current=(
            "initial_current",
            "mean"
        ),

        max_interrupted_load=(
            "interrupted_load_mwh",
            "max"
        ),


        # Operational activity
        switching_events=(
            "switch_device",
            "count"
        )

    )
    .reset_index()
)


print("Daily numerical dataset shape:")
print(daily_numeric.shape)

display(
    daily_numeric.head()
)
```

```python
# ==========================================
# Build Complete Daily Feeder Dataset
# ==========================================

import pandas as pd
from persiantools.jdatetime import JalaliDate


# ==========================================
# 1. Preserve feeder identity information
# ==========================================

# These were already confirmed to be constant
# within each MV feeder.

feeder_info = (
    df_reference
    .groupby("mv_feeder_gis")
    .agg(
        mv_feeder_code=("mv_feeder_code", "first"),
        mv_feeder=("mv_feeder", "first")
    )
    .reset_index()
)


# ==========================================
# 2. Preserve dynamic categorical/network data
# ==========================================

network_cols = [
    "region",
    "zone",
    "distribution_substation",
    "distribution_gis",
    "substation_type",
    "voltage_level",
    "lv_feeder"
]


# Keep every unique value observed in a feeder-day
# without one-hot / multi-hot encoding.

def combine_unique_values(series):

    values = (
        series
        .dropna()
        .astype(str)
        .str.strip()
    )

    values = [
        value
        for value in pd.unique(values)
        if value != ""
    ]

    if len(values) == 0:
        return pd.NA

    return " | ".join(values)


daily_network = (
    df_reference
    .groupby(
        [
            "mv_feeder_gis",
            "date"
        ]
    )[network_cols]
    .agg(combine_unique_values)
    .reset_index()
)


print("Daily network dataset shape:")
print(daily_network.shape)

display(daily_network.head())
```

```python
# ============================================================
# FINAL STEP: Build and Save Complete Daily Feeder Dataset
# ============================================================

import pandas as pd
from persiantools.jdatetime import JalaliDate


# ------------------------------------------------------------
# 1. Preserve static feeder identity
# ------------------------------------------------------------

feeder_info = (
    df_reference
    .groupby("mv_feeder_gis")
    .agg(
        mv_feeder_code=("mv_feeder_code", "first"),
        mv_feeder=("mv_feeder", "first")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Merge numerical + categorical/network information
# ------------------------------------------------------------

observed_daily = (
    daily_numeric
    .merge(
        daily_network,
        on=["mv_feeder_gis", "date"],
        how="left",
        validate="one_to_one"
    )
)


print("Observed feeder-days:")
print(observed_daily.shape)


# ------------------------------------------------------------
# 3. Create complete daily timeline for each feeder
# ------------------------------------------------------------

calendar_parts = []


for feeder, feeder_data in df_reference.groupby("mv_feeder_gis"):

    start_date = feeder_data["date"].min()
    end_date = feeder_data["date"].max()

    feeder_calendar = pd.DataFrame({
        "mv_feeder_gis": feeder,
        "date": pd.date_range(
            start=start_date,
            end=end_date,
            freq="D"
        )
    })

    calendar_parts.append(feeder_calendar)


full_calendar = pd.concat(
    calendar_parts,
    ignore_index=True
)


print("\nComplete calendar rows:")
print(full_calendar.shape)


# ------------------------------------------------------------
# 4. Merge observed data into complete calendar
# ------------------------------------------------------------

daily_health = (
    full_calendar
    .merge(
        observed_daily,
        on=["mv_feeder_gis", "date"],
        how="left",
        validate="one_to_one"
    )
    .merge(
        feeder_info,
        on="mv_feeder_gis",
        how="left",
        validate="many_to_one"
    )
)


# ------------------------------------------------------------
# 5. Identify true no-event days
# ------------------------------------------------------------

no_event_day = (
    daily_health["total_outages"].isna()
)


# These variables genuinely equal zero
# when there was no outage event.

zero_on_no_event_cols = [
    "unplanned_failures",
    "total_outages",
    "planned_outages",

    "total_duration",
    "unplanned_duration",
    "planned_duration",
    "max_duration",

    "total_energy_loss",
    "unplanned_energy_loss",
    "planned_energy_loss",
    "max_energy_loss",

    "max_interrupted_load",
    "switching_events"
]


daily_health.loc[
    no_event_day,
    zero_on_no_event_cols
] = 0


# Do NOT fill average_current.
# Missing means no outage-related current measurement.


# ------------------------------------------------------------
# 6. Restore integer count columns
# ------------------------------------------------------------

count_cols = [
    "unplanned_failures",
    "total_outages",
    "planned_outages",
    "switching_events"
]


daily_health[count_cols] = (
    daily_health[count_cols]
    .astype(int)
)


# ------------------------------------------------------------
# 7. Create daily binary failure indicator
# ------------------------------------------------------------

daily_health["failure_binary"] = (
    daily_health["unplanned_failures"] > 0
).astype(int)


# ------------------------------------------------------------
# 8. Gregorian calendar features
# ------------------------------------------------------------

daily_health["gregorian_year"] = (
    daily_health["date"].dt.year
)

daily_health["gregorian_month"] = (
    daily_health["date"].dt.month
)

daily_health["gregorian_day"] = (
    daily_health["date"].dt.day
)


# Pandas:
# Monday = 0 ... Sunday = 6

daily_health["weekday"] = (
    daily_health["date"].dt.weekday
)


# Iranian calendar week:
# Saturday = 0 ... Friday = 6

daily_health["iran_weekday"] = (
    (daily_health["weekday"] + 2) % 7
)


iran_weekday_names = {
    0: "Saturday",
    1: "Sunday",
    2: "Monday",
    3: "Tuesday",
    4: "Wednesday",
    5: "Thursday",
    6: "Friday"
}


daily_health["iran_weekday_name"] = (
    daily_health["iran_weekday"]
    .map(iran_weekday_names)
)


# ------------------------------------------------------------
# 9. Jalali calendar features
# ------------------------------------------------------------

def gregorian_to_jalali(date):

    try:

        date = pd.Timestamp(date)

        try:
            jdate = JalaliDate.to_jalali(
                date.year,
                date.month,
                date.day
            )

        except TypeError:
            jdate = JalaliDate.to_jalali(
                date.date()
            )

        return pd.Series([
            jdate.year,
            jdate.month,
            jdate.day
        ])

    except Exception:

        return pd.Series([
            pd.NA,
            pd.NA,
            pd.NA
        ])


daily_health[
    [
        "jalali_year",
        "jalali_month",
        "jalali_day"
    ]
] = (
    daily_health["date"]
    .apply(gregorian_to_jalali)
)


def get_jalali_season(month):

    if month in [1, 2, 3]:
        return "Spring"

    elif month in [4, 5, 6]:
        return "Summer"

    elif month in [7, 8, 9]:
        return "Autumn"

    elif month in [10, 11, 12]:
        return "Winter"

    return "Unknown"


daily_health["jalali_season"] = (
    daily_health["jalali_month"]
    .apply(get_jalali_season)
)


# ------------------------------------------------------------
# 10. Sort chronologically
# ------------------------------------------------------------

daily_health = (
    daily_health
    .sort_values(
        [
            "mv_feeder_gis",
            "date"
        ]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 11. Reorder important columns
# ------------------------------------------------------------

priority_cols = [
    "date",
    "mv_feeder_gis",
    "mv_feeder_code",
    "mv_feeder",

    "unplanned_failures",
    "failure_binary",

    "total_outages",
    "planned_outages"
]


remaining_cols = [
    col for col in daily_health.columns
    if col not in priority_cols
]


daily_health = daily_health[
    priority_cols + remaining_cols
]


# ------------------------------------------------------------
# 12. Final validation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL DAILY DATASET CHECK")
print("=" * 60)


print("\nFinal shape:")
print(daily_health.shape)


print("\nNumber of feeders:")
print(
    daily_health["mv_feeder_gis"].nunique()
)


print("\nDuplicate feeder-days:")
print(
    daily_health
    .duplicated(
        subset=[
            "mv_feeder_gis",
            "date"
        ]
    )
    .sum()
)


print("\nDays per feeder:")
display(
    daily_health
    .groupby("mv_feeder_gis")["date"]
    .count()
)


print("\nNo-event days:")
print(
    (
        daily_health["total_outages"] == 0
    ).sum()
)


print("\nFailure days:")
print(
    daily_health["failure_binary"].sum()
)


print("\nFailure binary distribution:")
display(
    daily_health["failure_binary"]
    .value_counts()
    .sort_index()
)


print("\nDate range:")
print(
    daily_health["date"].min(),
    "to",
    daily_health["date"].max()
)


network_cols = [
    "region",
    "zone",
    "distribution_substation",
    "distribution_gis",
    "substation_type",
    "voltage_level",
    "lv_feeder"
]


print("\nNetwork columns preserved:")

for col in network_cols:

    print(
        col,
        "✓" if col in daily_health.columns
        else "MISSING"
    )


print("\nMissing Jalali dates:")
print(
    daily_health["jalali_year"]
    .isna()
    .sum()
)


display(
    daily_health.head(10)
)


# ------------------------------------------------------------
# 13. Save final CSV
# ------------------------------------------------------------

output_path = "/content/daily_feeder_health(II).csv"


daily_health.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig",
    date_format="%Y-%m-%d"
)


print("\nSaved successfully:")
print(output_path)
```

```python
# ============================================================
# FINAL COLUMN AUDIT
# Original Dataset vs Final Daily Dataset
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# 1. Normalize column names for reliable matching
# ------------------------------------------------------------

def normalize_name(name):

    name = str(name)

    return (
        name
        .replace("ي", "ی")
        .replace("ى", "ی")
        .replace("ك", "ک")
        .replace("ۀ", "ه")
        .strip()
    )


def normalize_dict_keys(dictionary):

    return {
        normalize_name(key): value
        for key, value in dictionary.items()
    }


# ------------------------------------------------------------
# 2. Original → preserved final columns
# ------------------------------------------------------------

preserved_map = normalize_dict_keys({

    "منطقه اصلی": "region",

    "فیدر فشار متوسط": "mv_feeder",
    "کد فیدر فشار متوسط": "mv_feeder_code",
    "GISCODE فیدر فشار متوسط": "mv_feeder_gis",

    "پست توزیع": "distribution_substation",
    "GISCODE پست توزیع": "distribution_gis",

    "فیدر فشار ضعیف": "lv_feeder",
    "سطح ولتاژ": "voltage_level",

    "نوع پست": "substation_type",
    "زون": "zone"
})


# ------------------------------------------------------------
# 3. Original → transformed / aggregated columns
# ------------------------------------------------------------

transformed_map = normalize_dict_keys({

    "تاریخ اطلاع": (
        "date + calendar features",
        "Converted to daily Gregorian/Jalali calendar information"
    ),

    "ساعت اطلاع": (
        "Not directly retained",
        "Used to construct event datetime before daily aggregation"
    ),

    "روز هفته": (
        "weekday, iran_weekday, iran_weekday_name",
        "Recreated from the final daily date"
    ),

    "نوع قطعی": (
        "unplanned_failures, planned_outages, failure_binary",
        "Used to distinguish planned and unplanned outages"
    ),

    "مدت زمان خاموشی به دقیقه": (
        "total_duration, unplanned_duration, planned_duration, max_duration",
        "Aggregated by feeder-day"
    ),

    "انرژی توزیع نشده (KWH)": (
        "total_energy_loss, unplanned_energy_loss, planned_energy_loss, max_energy_loss",
        "Aggregated by feeder-day"
    ),

    "جریان قطع اولیه": (
        "average_current",
        "Daily mean of available outage-current measurements"
    ),

    "بار قطع شده (MWH)": (
        "max_interrupted_load",
        "Daily maximum interrupted load"
    ),

    "جداکننده": (
        "switching_events",
        "Counted by feeder-day"
    )
})


# ------------------------------------------------------------
# 4. Original columns intentionally not carried forward
# ------------------------------------------------------------

not_carried_notes = normalize_dict_keys({

    "ردیف":
        "Administrative row identifier",

    "ردیف یکتا":
        "Event-level identifier",

    "آدرس":
        "High-cardinality event-level location text",

    "گروه قطع":
        "Event-level outage information not used in current daily feature set",

    "علت قطع":
        "Event-level outage reason not used in current daily feature set",

    "اشکال":
        "Event-level fault information not used in current daily feature set",

    "علت اشکال":
        "Event-level fault detail not used in current daily feature set",

    "مدت قطع":
        "Not used; outage_minutes was used for daily duration",

    "تکی یا کلی به تشخیص اپراتور":
        "Event-level outage scope",

    "شرح حادثه":
        "Free-text event description",

    "زمان قطع":
        "Event-level outage timestamp",

    "زمان وصل":
        "Event-level restoration timestamp",

    "كاربري كه ثبت كرده":
        "Administrative user information",

    "گزارش دهنده":
        "Reporting information",

    "کاربر ثبت کننده":
        "Administrative user information",

    "source_file":
        "Source-file provenance only"
})


# ------------------------------------------------------------
# 5. Detect columns actually removed during Step 2
# ------------------------------------------------------------

cleaned_names = {
    normalize_name(col)
    for col in df_clean.columns
}


removed_names = {
    normalize_name(col)
    for col in data.columns
    if normalize_name(col) not in cleaned_names
}


# Determine whether removed column was empty or constant
removed_reasons = {}


for col in data.columns:

    key = normalize_name(col)

    if key not in removed_names:
        continue

    series = data[col].copy()

    if series.dtype == "object":

        series = (
            series
            .astype("string")
            .str.strip()
            .replace(r"^\s*$", pd.NA, regex=True)
        )

    if series.isna().all():

        removed_reasons[key] = "Completely empty"

    elif series.nunique(dropna=True) <= 1:

        removed_reasons[key] = "Constant column"

    else:

        removed_reasons[key] = "Removed during basic cleaning"


# ------------------------------------------------------------
# 6. Build ORIGINAL column audit
# ------------------------------------------------------------

network_original_cols = {
    normalize_name(x)
    for x in [
        "منطقه اصلی",
        "پست توزیع",
        "GISCODE پست توزیع",
        "فیدر فشار ضعیف",
        "سطح ولتاژ",
        "نوع پست",
        "زون"
    ]
}


original_audit = []


for original_col in data.columns:

    key = normalize_name(original_col)


    if key in removed_names:

        status = "REMOVED"
        final_representation = "-"
        note = removed_reasons[key]


    elif key in preserved_map:

        status = "PRESERVED / RENAMED"
        final_representation = preserved_map[key]

        if key in network_original_cols:

            note = (
                "Preserved as daily network information; "
                "multiple same-day values may be combined"
            )

        else:

            note = "Preserved as feeder identity information"


    elif key in transformed_map:

        status = "TRANSFORMED / AGGREGATED"
        final_representation = transformed_map[key][0]
        note = transformed_map[key][1]


    elif key in not_carried_notes:

        status = "NOT CARRIED TO FINAL"
        final_representation = "-"
        note = not_carried_notes[key]


    else:

        status = "REVIEW"
        final_representation = "-"
        note = "No transformation rule found"


    original_audit.append({

        "Original Column": original_col,
        "Status": status,
        "Final Representation": final_representation,
        "Brief Explanation": note

    })


original_audit = pd.DataFrame(original_audit)


# ------------------------------------------------------------
# 7. Define lineage of every FINAL column
# ------------------------------------------------------------

final_lineage = {

    "date":
        ("DERIVED", "تاریخ اطلاع",
         "Gregorian daily date and complete calendar"),

    "mv_feeder_gis":
        ("PRESERVED / RENAMED", "GISCODE فیدر فشار متوسط",
         "Main feeder identifier"),

    "mv_feeder_code":
        ("PRESERVED / RENAMED", "کد فیدر فشار متوسط",
         "Static feeder code"),

    "mv_feeder":
        ("PRESERVED / RENAMED", "فیدر فشار متوسط",
         "Static feeder name"),

    "unplanned_failures":
        ("AGGREGATED", "نوع قطعی",
         "Number of unplanned outage events per feeder-day"),

    "failure_binary":
        ("DERIVED", "unplanned_failures",
         "1 if at least one unplanned failure occurred"),

    "total_outages":
        ("AGGREGATED", "All outage records",
         "Total outage events per feeder-day"),

    "planned_outages":
        ("AGGREGATED", "نوع قطعی",
         "Number of planned outages per feeder-day"),

    "total_duration":
        ("AGGREGATED", "مدت زمان خاموشی به دقیقه",
         "Total outage duration"),

    "unplanned_duration":
        ("AGGREGATED", "مدت زمان خاموشی به دقیقه + نوع قطعی",
         "Total unplanned outage duration"),

    "planned_duration":
        ("AGGREGATED", "مدت زمان خاموشی به دقیقه + نوع قطعی",
         "Total planned outage duration"),

    "max_duration":
        ("AGGREGATED", "مدت زمان خاموشی به دقیقه",
         "Maximum outage duration"),

    "total_energy_loss":
        ("AGGREGATED", "انرژی توزیع نشده (KWH)",
         "Total daily energy not supplied"),

    "unplanned_energy_loss":
        ("AGGREGATED", "انرژی توزیع نشده (KWH) + نوع قطعی",
         "Unplanned-outage energy loss"),

    "planned_energy_loss":
        ("AGGREGATED", "انرژی توزیع نشده (KWH) + نوع قطعی",
         "Planned-outage energy loss"),

    "max_energy_loss":
        ("AGGREGATED", "انرژی توزیع نشده (KWH)",
         "Maximum event energy loss"),

    "average_current":
        ("AGGREGATED", "جریان قطع اولیه",
         "Mean outage-related current"),

    "max_interrupted_load":
        ("AGGREGATED", "بار قطع شده (MWH)",
         "Maximum interrupted load"),

    "switching_events":
        ("AGGREGATED", "جداکننده",
         "Number of non-missing switching-device records"),

    "region":
        ("PRESERVED / DAILY", "منطقه اصلی",
         "Region values observed during feeder-day"),

    "zone":
        ("PRESERVED / DAILY", "زون",
         "Zone values observed during feeder-day"),

    "distribution_substation":
        ("PRESERVED / DAILY", "پست توزیع",
         "Distribution substations observed during feeder-day"),

    "distribution_gis":
        ("PRESERVED / DAILY", "GISCODE پست توزیع",
         "Distribution GIS assets observed during feeder-day"),

    "substation_type":
        ("PRESERVED / DAILY", "نوع پست",
         "Substation types observed during feeder-day"),

    "voltage_level":
        ("PRESERVED / DAILY", "سطح ولتاژ",
         "Voltage levels observed during feeder-day"),

    "lv_feeder":
        ("PRESERVED / DAILY", "فیدر فشار ضعیف",
         "LV feeders observed during feeder-day"),

    "gregorian_year":
        ("DERIVED", "date", "Gregorian year"),

    "gregorian_month":
        ("DERIVED", "date", "Gregorian month"),

    "gregorian_day":
        ("DERIVED", "date", "Gregorian day"),

    "weekday":
        ("DERIVED", "date", "Pandas weekday: Monday=0"),

    "iran_weekday":
        ("DERIVED", "date", "Iranian weekday: Saturday=0"),

    "iran_weekday_name":
        ("DERIVED", "date", "Iranian weekday name"),

    "jalali_year":
        ("DERIVED", "date", "Jalali year"),

    "jalali_month":
        ("DERIVED", "date", "Jalali month"),

    "jalali_day":
        ("DERIVED", "date", "Jalali day"),

    "jalali_season":
        ("DERIVED", "jalali_month", "Jalali season")
}


# ------------------------------------------------------------
# 8. Build FINAL column audit
# ------------------------------------------------------------

final_audit = []


for col in daily_health.columns:

    if col in final_lineage:

        status, source, explanation = final_lineage[col]

    else:

        status = "REVIEW"
        source = "Unknown"
        explanation = "No lineage rule found"


    final_audit.append({

        "Final Column": col,
        "Status": status,
        "Source": source,
        "Brief Explanation": explanation

    })


final_audit = pd.DataFrame(final_audit)


# ------------------------------------------------------------
# 9. Display complete audits
# ------------------------------------------------------------

print("=" * 80)
print("ORIGINAL DATASET COLUMN AUDIT")
print("=" * 80)

print(f"\nOriginal columns: {len(data.columns)}")

display(original_audit)


print("\n" + "=" * 80)
print("FINAL DAILY DATASET COLUMN AUDIT")
print("=" * 80)

print(f"\nFinal columns: {len(daily_health.columns)}")

display(final_audit)


# ------------------------------------------------------------
# 10. Status summaries
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("ORIGINAL COLUMN STATUS SUMMARY")
print("=" * 80)

display(
    original_audit["Status"]
    .value_counts()
    .rename_axis("Status")
    .reset_index(name="Number of Columns")
)


print("\n" + "=" * 80)
print("FINAL COLUMN STATUS SUMMARY")
print("=" * 80)

display(
    final_audit["Status"]
    .value_counts()
    .rename_axis("Status")
    .reset_index(name="Number of Columns")
)


# ------------------------------------------------------------
# 11. Final safety check
# ------------------------------------------------------------

original_review = (
    original_audit["Status"]
    .eq("REVIEW")
    .sum()
)

final_review = (
    final_audit["Status"]
    .eq("REVIEW")
    .sum()
)


print("\n" + "=" * 80)

if original_review == 0 and final_review == 0:

    print("COLUMN AUDIT COMPLETE ✓")
    print("Every original and final column has been accounted for.")

else:

    print("COLUMN AUDIT NEEDS REVIEW ⚠")
    print("Unclassified original columns:", original_review)
    print("Unclassified final columns:", final_review)

print("=" * 80)
```
